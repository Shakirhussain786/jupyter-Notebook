```python
def extract(txt): # this is a function which will seprate only inside the bracket values mean here the value is epsode number
    check = False
    data = ""
    for ch in txt:
        if ch == ")":
            check = False
            break
            return data
        if check == True:
            data = data + ch
        if ch == "(":
            check = True
            
    return data


```

```python
df["title"].apply(extract)
```

```python
df["Epesode"] = df["title"].apply(extract)
df
```

```python
df["Epesode"] = df["Epesode"].str.replace("ep ","")
df
```

```python
df["Epesode"] = df["Epesode"].astype(int) #use to convert str value into int type
df
```

```python
def extract_time(dat):
    check = False
    time = ""
    count = 0
    for ch in dat:
        if ch == ")":
            check = True
        if check == True:
            time = time + ch
            count += 1
        if count == 19:
            check = False
            break
                
    return time
    
        
```

```python
df["Time stampe"] = df["title"].apply(extract_time)
df
```
