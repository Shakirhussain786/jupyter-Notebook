```python
plt.Figure(figsize=(10,5))
plt.title("population vs year chart")
plt.style.use("bmh")
plt.xlabel("population")
plt.ylabel("years")
plt.xticks([0,5,9],labels=["small","medium","large"])
plt.plot(x,y,linestyle="_.",lable="x**2",color="red")
plt.plot(x,x**3,lable="x**3",linestyle="--",linewidth=3)
plt.plot(x,x**4,lable="x**4")
plt.ylim(0,1000)
plt.grid()
plt.legend()

```
