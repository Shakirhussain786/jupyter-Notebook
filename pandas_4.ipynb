```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "date" : pd.date_range("2026-4-24",periods=20),
    "product" : ['A','B','C','D'] * 5,
    "region" : ["east","west","south","north"] *5,
    "sales" : np.random.randint(100,1000,20),
    "unites" : np.random.randint(10,100,20),
    "rep" : ["jhon","bob","david","max"] *5
}

df = pd.DataFrame(data)
df
```

```python
df["month"] = df["date"].dt.month_name()
df["qourter"] = "Q" + df["date"].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values= "sales",index= "region",columns= "product",)
```
