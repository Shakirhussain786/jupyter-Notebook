```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.impute import SimpleImputer
num_data = x_train.drop("ocean_proximity",axis=1)
cat_data = x_train[["ocean_proximity"]]

imp = SimpleImputer(strategy="mean")

num_impute_data = imp.fit_transform(num_data)
num_df = pd.DataFrame(num_impute_data,columns=imp.get_feature_names_out(),index=num_data.index)
num_df.head()
```
