```python
import numpy as np 
import pandas as pd

data = {
    "name" : ["shair",np.nan,"ali","ahmad","hassan"],
    "gender" : ["male",np.nan,"male","femail","femail"],
    "age" : [12,16,19,np.nan,20],
    "city" : ["karachi","gb",np.nan,"panjab","skardu"]
}

df = pd.DataFrame(data,index=[1,2,3,4,5])
```

```python
print(df)
```

```python
df.isna()
```

```python
df.isna().sum() #isna mean null or missiny value and .sum mean how many null value in a single row
```

```python
df.isna().any() #it will check every colonms if there is an any missing or null value it will print true elsse 
#else if there is no any null value it will print false
```

```python
df
```

```python
df.dropna() # it will check every row if any null value in any row it will remove the row in this case you can see
```

```python
df
```

```python
df.dropna(thresh=4) # thresh mean atleast 4 value is correect than it will print the rows that have atleast 4 true value
```

```python
df
```

```python
df.fillna(0) #fillna will fill 0 where null value 
```
